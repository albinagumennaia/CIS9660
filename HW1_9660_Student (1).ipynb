## Part B: Data Preparation

### Import libraries

```python
from sklearn.impute import SimpleImputer             # for handling missing data
from sklearn.preprocessing import LabelEncoder       # for encoding label data
from sklearn.preprocessing import StandardScaler     # for standardizing the data
```

### Load data
* Independent variables: Region, Age, Income
* Dependent variable: Online_Shopper

3.5 Please read data file (shoppers.csv) into a dataframe

```python
df_shoppers = pd.read_csv('https://raw.githubusercontent.com/vjavaly/Baruch-CIS-9660/main/data/shoppers.csv')
```

3.6 Please check missing values in this shopper dataset

```python
df_shoppers.isnull().sum()
```

3.7 Please use the SimpleImputer to replace ALL missing values (strategy="mean")

```python
# Setting up SimpleImputer with "mean"
imp_mean = SimpleImputer(missing_values = np.nan, strategy = 'mean')
```

```python
# Apllying the imputer the columns with missing values and saving it back into df_shoppers
df_shoppers[['Age', 'Income']]=imp_mean.fit_transform(df_shoppers[['Age', 'Income']])
df_shoppers.isnull().sum()
```

3.8 Generate one-hot encodings for Region column

```python
# One-hot encoding for Region column
df_shoppers = pd.get_dummies(df_shoppers, prefix = ['Region'], columns = ['Region'])
df_shoppers.head()
```

```python
# Converting booleans to integers
df_shoppers[['Region_Brazil', 'Region_India', 'Region_USA']] = \
df_shoppers[['Region_Brazil', 'Region_India', 'Region_USA']].astype(int)
df_shoppers.head()
```

3.9 Please use dummy variables to Encode labels for Online_Shopper and combine dummies into the original dataframe

```python
# Dummy encoding for 'Online_Shopper' column, converting to integers right away and dropping first column 'No'
df_shoppers = pd.get_dummies(df_shoppers, columns =['Online_Shopper'], drop_first = True, dtype = int)
df_shoppers.head()
```

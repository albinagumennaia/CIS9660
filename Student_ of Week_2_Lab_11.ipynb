### Import libraries

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
```

### Load data
#### Original data file from Kaggle: https://www.kaggle.com/uciml/electric-power-consumption-data-set/metadata
#### Note: A smaller version of the original file is in GitHub: household_power_consumption.txt

```python
# Load data file (household_power_consumption.txt) into dataframe
#  NOTE: field separator is ';'
df = pd.read_csv('https://raw.githubusercontent.com/vjavaly/Baruch-CIS-STA-3920/main/data/household_power_consumption.txt',
                 sep=';', low_memory=False)
df.shape
```

### Prepare an even smaller dataset

```python
# From large dataset generate a smaller dataset (a random sample of 10,000 records)
df3 = df.sample(n=10000, random_state=42)
df3.shape
```

```python
# Review first few records in dataframe
df3.head()
```

### Data imputation

```python
# Check for null values/missing values
df3.isnull().sum()
```

```python
# Identify rows with missing values
df3[df3.isnull().any(axis=1)]
```

#### Note: Though 135 rows have '?' in multiple columns, those columns are not listed in 'isnull().sum()' output

```python
# Replace '?' with NaN across all columns
df3['Global_active_power'].replace('?', np.nan, inplace=True)
df3['Global_reactive_power'].replace('?', np.nan, inplace=True)
df3['Voltage'].replace('?', np.nan, inplace=True)
df3['Global_intensity'].replace('?', np.nan, inplace=True)
df3['Sub_metering_1'].replace('?', np.nan, inplace=True)
df3['Sub_metering_2'].replace('?', np.nan, inplace=True)
```

```python
# Check for null values
df3.isnull().sum()
```

```python
# Verify cells with '?' have been replaced with NaN
df3[df3.isnull().any(axis=1)]
```

```python
# Setup imputer to replace NaN cells with mean of column
#  Both hyperparameters explicitly specified for teaching purposes
imp_mean = SimpleImputer(missing_values=np.nan, strategy='mean')
```

```python
# The fit method computes the necessary statistics (e.g. mean, median, mode, contant)
tmp_1 = df3[['Global_active_power', 'Global_reactive_power', 'Voltage', 'Global_intensity',
             'Sub_metering_1', 'Sub_metering_2', 'Sub_metering_3' ]]
imp_mean.fit(tmp_1)
```

```python
# The transform method applies the imputation to the data by replacing the missing values
#  with the computed statistics
tmp_2 = imp_mean.transform(tmp_1)
```

```python
# The fit_transform method is a convenience method that combines the functionality of
#  the fit and transform methods
tmp_3 = imp_mean.fit_transform(tmp_1)
```

```python
np.array_equal(tmp_2, tmp_3)
```

```python
tmp_3
```

```python
# Convert the transformed data into a dataframe
df4 = pd.DataFrame(tmp_3,
                   columns=['Global_active_power', 'Global_reactive_power', 'Voltage',
                            'Global_intensity', 'Sub_metering_1', 'Sub_metering_2', 'Sub_metering_3'])
df4.head(10)
```
